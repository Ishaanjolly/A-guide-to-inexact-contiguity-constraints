Tests 4×4 and 6×6 grids across a range of population deviations.

```python
import sys
from pathlib import Path

ROOT_DIR = Path.cwd().parents[1]
sys.path.insert(0, str(ROOT_DIR))
```

```python
import os
import networkx as nx
import pandas as pd

from src.mip import multi_district_mip
from src.utils import make_grid_graph

os.makedirs("../results", exist_ok=True)
```

##  SET EXPERIMENT PARAMETERS

```python
DEVIATIONS = [0, 1, 2]
CONTIGUITY = ["dag"]
POOL_SIZE = 1000000
TIME_LIMIT = 3600
```

## 4×4 Grid

```python
G4 = make_grid_graph(4, 4)


print(f"4×4 grid: {G4.number_of_nodes()} nodes, {G4.number_of_edges()} edges")
```

```python
nx.draw(
    G4,
    pos={i: (G4.nodes[i]["X"], G4.nodes[i]["Y"]) for i in G4.nodes},
    with_labels=True,
)
```

```python
## ROOT SELECTION
roots = [0, 3, 12, 15]
```

```python
results_4x4 = []

for deviation in DEVIATIONS:
    for contiguity in CONTIGUITY:
        print(f"\ndeviation={deviation} | contiguity={contiguity}")

        metrics, _ = multi_district_mip(
            G=G4,
            deviation_persons=deviation,
            contiguity=contiguity,
            roots=roots,
            k=len(roots),
            pool_size=POOL_SIZE,
            pool_search=2,
            time_limit=TIME_LIMIT,
        )
        metrics.update(
            {"grid": "4x4", "deviation": deviation, "contiguity": contiguity}
        )

        results_4x4.append(metrics)

df_4x4 = pd.DataFrame(results_4x4)
df_4x4.to_csv("../../results/small_4x4_enumeration.csv", index=False)
```
