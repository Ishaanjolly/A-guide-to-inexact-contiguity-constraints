## Iowa Optimization — Hop M (Unweighted Distances)

This notebook runs multi-district MIP optimization experiments on the Iowa county graph using **hop-based distances** (unweighted BFS) for the tree, dist, and dag contiguity constraints.

**Hop M** means the Big-M parameter in the distance-based contiguity formulations is derived from hop distances: every edge has weight 1, so the shortest path between two nodes is simply the number of hops. This mirrors the hop-distance version of the feasibility analysis.

### Experiment parameters

We sweep over a range of population deviations, all contiguity formulations, and a representative set of objectives. `distance_metric=None` means hop distances are used — edges are unweighted (each hop costs 1).

```python
DEVIATIONS   = [50, 100, 200, 500, 1000]
CONTIGUITY   = ["tree", "dist", "dag", "cut"]
OBJECTIVES   = ["hop_moi", "euclidean_moi", "cut_edges"]
TIME_LIMIT   = 3600

total = len(DEVIATIONS) * len(CONTIGUITY) * len(OBJECTIVES)
print(f"Deviations : {DEVIATIONS}")
print(f"Contiguity : {CONTIGUITY}")
print(f"Objectives : {OBJECTIVES}")
print(f"Total experiments to run: {total}")
```

```python
df = run_optimization_experiment(
    G_base=G,
    deviations=DEVIATIONS,
    contiguity_models=CONTIGUITY,
    objectives=OBJECTIVES,
    k=K,
    distance_metric=None,  # Hop M: unweighted BFS distances
    time_limit=TIME_LIMIT,
    results_file="../results/IA_county_optimization_hop_M.csv",
)
```
