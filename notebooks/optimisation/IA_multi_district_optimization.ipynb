```python
import sys
from pathlib import Path


ROOT_DIR = Path.cwd().parents[1]
sys.path.insert(0, str(ROOT_DIR))

import os
import matplotlib.pyplot as plt

from src.read import read_graph_from_json
from src.utils import get_roots
from src.experiment import run_optimization_experiment
from src.draw import visualize_roots


os.makedirs("../../results", exist_ok=True)
```

```python
G_ia = read_graph_from_json("../../data/IA_county.json", state="IA")


print(
    f"Iowa county graph: {G_ia.number_of_nodes()} nodes, {G_ia.number_of_edges()} edges"
)
```

```python
K = 4
ia_roots = get_roots(G_ia, k=K, root_strategy="corner")
print(f"Roots: {ia_roots}")

fig = visualize_roots(G_ia, ia_roots, title="Iowa Counties — Optimisation Roots (k=4)")
plt.show()
```

```python
DEVIATIONS = [10]
CONTIGUITY = ["dist"]
OBJECTIVES = [
    "hop_moi",
    "weighted_moi",
    "euclidean_moi",
    "cut_edges",
    "shared_perim",
    "perim",
    "inverse_polsby_popper",
]
TIME_LIMIT = 3600

total = len(DEVIATIONS) * len(CONTIGUITY) * len(OBJECTIVES)
print(f"Deviations : {DEVIATIONS}")
print(f"Contiguity : {CONTIGUITY}")
print(f"Objectives : {OBJECTIVES}")
print(f"Total experiments to runs : {total}")
```

```python
df = run_optimization_experiment(
    G_base=G_ia,
    deviations=DEVIATIONS,
    contiguity_models=CONTIGUITY,
    objectives=OBJECTIVES,
    k=K,
    distance_metric="euclidean",
    time_limit=TIME_LIMIT,
    root_strategy="corner",
    results_file="../results/IA_county_optimisation.csv",
)
```
